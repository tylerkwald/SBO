```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("withAFW.dat")
```

```python
#Inlet Temperature Plot
plt.figure()
plt.clf
plt.plot(df["time"], df['afw_intact_flow'], label = 'afw')
#plt.plot(df["time"], df['br_mfw_flow1'], label = 'mfw')


plt.xlabel('Time (s)')
plt.ylabel('Flow Rate (kg/s)')
plt.title('AFW Flow')
plt.legend()
#plt.savefig('graphs/AFW_flow_wafw.png')

```

```python
#Peak Cladding Temperature

df["pct"] = df['core_temp_1']
for i in range(2,13):
    df['pct'] = df[['pct', 'core_temp_'+str(i)]].max(axis=1)

plt.figure()
plt.clf
for i in range(1,13):
    plt.plot(df["time"], df['core_temp_'+str(i)], alpha=0.4, label='Node '+str(i))
plt.plot(df["time"], df['pct'], label = 'PCT', color = 'red')

#plt.xlim([500, 12000])
#plt.xlim([495, 12000])
plt.xlabel('Time (s)')
plt.ylabel('PCT (K)')
plt.title('Peak Cladding Temperature')
plt.savefig('graphs/pct_wafw.png')
plt.legend()
```
